```python
# [{},{},{}]
# list 안에 (같은 키를 가지고 있는) dictionary 여러개

orders = [
    {"product": "키보드", "price": 50000, "quantity": 2},
    {"product": "마우스", "price": 30000, "quantity": 1},
    {"product": "모니터", "price": 250000, "quantity": 1},
    {"product": "키보드", "price": 50000, "quantity": 1}
]

# 1. 전체 매출
# 2. 주문  건수
# 3. 가장 비싼 주문
# 4. 10만원 이상 주문내역

total = 0
max_price = 0
count = 0

for order in orders:
    amount = order['price'] * order['quantity'] # 현 주문의 매출
    total += amount # 전체 매출 

    # 최대 주문금액
    if max_price < amount:
        max_price = amount

    # 10만원 이상이면 cout + 1

    if amount >= 100000:
        count += 1

print('전체 매출:', total)
print('주문 건수:', len(orders))
print('최대 주문 금액:', max_price)
print('10만원 이상 건수:', count)        
```

```python
student = {
      "name" : "홍길동",
      "class" : "A반", 
      "scores" : [85,90,78,92,88]
}

# 학생의 이름
print('이름:', student['name'])
# 반
print('반:', student['class'])
# 평균 출력 합계sum()/갯수len()
print('평균', sum(student['scores']) / len(student['scores']))
```

```python
employees = [
    {
        "name": "김대리",
        "department": "개발팀",
        "skills": ["Python", "SQL", "React"]
    },
    {
        "name": "이과장",
        "department": "기획팀",
        "skills": ["Excel", "PowerPoint", "SQL"]
    },
    {
        "name": "박사원",
        "department": "개발팀",
        "skills": ["Java", "Spring", "Oracle"]
    },
    {
        "name": "최대리",
        "department": "AI팀",
        "skills": ["Python", "TensorFlow", "PyTorch"]
    },
    {
        "name": "정사원",
        "department": "개발팀",
        "skills": ["JavaScript", "React", "Node.js"]
    }
]

print("Python 개발 가능 직원")


for employee in employees:
    if "Python" in employee["skills"]:
        print(f"이름: {employee['name']}, 팀: {employee['department']}")


for employee in employees:
    if "Python" in employee["skills"]:
        print(employee['name'], '-', employee['department'])
```
