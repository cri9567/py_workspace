```python
# set -> 중복 없음(unique)
# 순서 X -> index 없음

# 선언
# 저장 시점에 중복값 걸러냄
# 저장되는 순서는 유지되지 않는다.
s = {1,2,3,'1',2,4}
print(s, type(s))
```

```python
print(s[0]) # 인덱싱 불가
```

```python
for i in s:
    print(i)
```

```python
# 추가
s.add(5) # 없으면 추가
print(s)

s.add(1) # 있으면 추가되지 않음
print(s)
```

```python
# 삭제
s = {1,2,3,4}
s.remove(1) #없는 값 삭제시 error
print(s)

if 10 in s:
    print('10 있으면 삭제')
    s.remove(10)
else:
    print('10 없음')    
```

```python
s.discard(10) # 삭제(없는 값 삭제해도 error 안남)
print(s)
```

```python
s.update([1,3,5,7,9]) # 기존 set에 여러 값을 한번에 추가 
print(s)
```
