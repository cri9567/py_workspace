````python
가전제품의 공통 기능을 부모 클래스인 `Appliance`에 작성하고, 이를 상속받는 `TV` 클래스를 구현하시오.

### 1. 부모 클래스 `Appliance` 작성

다음 조건에 맞게 클래스를 작성하시오.

1.  `__init__()`는 가전제품의 브랜드를 매개변수로 전달받는다.
2. 브랜드는 공개 속성 `brand`에 저장한다.
3. 전원 상태는 비공개 속성 `__power`에 저장하며, 처음에는 `False`로 설정한다.
4. `turn_on_off()` 메서드는 전원 상태를 반대로 변경한다.
   - 전원이 켜지면 `전원이 켜졌습니다.` 와 같이 출력한다.
   - 전원이 꺼지면 `전원이 꺼졌습니다.` 와 같이 출력한다.
5. `is_power_on()` 메서드는 현재 전원 상태를 반환한다. 외부에서는 이 메서드로 전원 상태를 확인할 수 있지만, 전원 상태를 직접 변경할 수는 없도록 한다.

### 2. 자식 클래스 `TV` 작성

`Appliance`를 상속받는 `TV` 클래스를 작성하시오.

1. __init__() 는 브랜드를 전달받는다.
2. 채널의 기본값은 `1`로 설정한다.
3. 부모 클래스의 생성자를 `super()`로 호출한다.
4. `channel` 속성에 현재 채널을 저장한다.
5. `set_channel(ch)` 메서드를 작성한다.
   - 전원이 켜져 있으면 채널을 `ch`로 변경하고 `7번 채널로 변경되었습니다.` 와 같이 출력한다.
   - 전원이 꺼져 있으면 채널을 변경하지 않고 `전원을 먼저 켜주세요.` 를 출력한다.
   - 전원 상태는 부모 클래스(super())의 `is_power_on()` 메서드를 이용해 확인한다.

### 3. 실행 결과

아래 테스트 코드를 실행했을 때 예상 결과와 동일하게 출력되도록 작성하시오.

```python
my_tv = TV("삼성")

my_tv.set_channel(7)

my_tv.turn_on_off()
my_tv.set_channel(7)

my_tv.turn_on_off()
my_tv.set_channel(11)
```

예상 출력:

```text
전원을 먼저 켜주세요.
전원이 켜졌습니다.
7번 채널로 변경되었습니다.
전원이 꺼졌습니다.
전원을 먼저 켜주세요.
```
````

````python
### 1. 부모 클래스 `Appliance` 작성

class Appliance:
    def __init__(self, b):
        self.brand = b
        self.__power = False

    def turn_on_off(self):
        self.__power = not self.__power

        if self.__power:
            print('전원이 켜졌습니다.')
        else:
            print('전원이 꺼졌습니다.')
    def is_power_on(self):
        return self.__power    



### 2. 자식 클래스 `TV` 작성

class TV(Appliance):
    def __init__(self, b):
        super().__init__(b)
        self.channel = 1

    def set_channel(self, ch):
        if super().is_power_on():
            self.channel = ch
            print(f'{ch}번 채널로 변경되었습니다.')
        else:
            print('전원을 먼저 켜주세요.') 

my_tv = TV("삼성")

my_tv.set_channel(7)

my_tv.turn_on_off()
my_tv.set_channel(7)

my_tv.turn_on_off()
my_tv.set_channel(11)                                       

````
