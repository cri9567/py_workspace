```python
# f'' : 변수에 저장된 값을 꺼내와 {}에 넣어준다(값 직접 지정해야 됨)
# 변수 : 하나의 값을 저장하는 공간
name = '양인준'
age = 37
print(f'안녕 내 이름은 {name}이고 나이는 {age}살이고 직업은 {"기사"}야')
# {변수명}, {값}, : 출력할 구조가 한눈에 보임
```

```python
# 동일한 틀에 데이터만 계속 바꿔 들어가는 시스템을 만들때 포맷팅을 사용한다.(로그 찍기 등)
```

```python
print(f"{7:3}")
print(f"{7:05}")
money = 1234567
print(f"{money:,}원")
print(f"{'번호':<6}{'이름':<10}{'점수':^6}")
print(f"{1:<6}{'홍길동':<10}{'95':>6}")
print(f"{1:<6}{'이홍동':<10}{'105':>6}")
# <:왼쪽 정렬 >:오른쪽 정렬 ^:가운데 정렬
```
