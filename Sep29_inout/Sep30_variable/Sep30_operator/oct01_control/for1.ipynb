```python
# 반복문
# for : 지정된 횟수만큼 수행문을 반복 실행
# while : 조건에 만족하는 동안(True) 반복 실행
```

```python
'''
for 변수 in 범위
    반복 수행할 문장
    반복 수행할 문장...

범위 : range(), 시퀀스 자료형(문자열, 리스트, 튜플...)


'''
```

```python
# in 문자열 -> 문자 갯수만큼 반복
for i in '가나다':
    print(i) # 5번 수행
```

```python
# 0 ~ 19까지 다 출력하세요
# range() : 범위를 만들어주는 함수
# range(end) ->시작 0, 끝 end -1, 증가 1
for i in range(10): # 0 ~ 9
    print(i, end =' ')
print() # 줄바꿈
for i in range(100): 
    print(i, end =' ')
```

```python
# range(start, end) -> 시작 start, 끝 end-1, 증가 1
for i in range(1, 9): # 
    print(i, end=' ')
```

```python
# range(start, and, step) -> 시작 strt, 끝 end-1, 증가 step
for i in range(1, 50, 10):
    print(i, end=' ')
```

```python
# 'hello' 3번 출력하세요

# 변수에 저장된 값을 수행문에서 사용하지 않는다면
# 굳이 변수명을 지정하지 않고 변수명을 _로 작성(권장)
```

```python
for _ in range(3):
    print('hello')
    print(_)
print('for 문 끝나고 나서', _)    

  
    
```

```python
# 1 ~ 100까지의 합을 구하세요
```

```python
sum_value = 0 # 합계를 저장할 공간

for i in range(1, 101):
    sum_value += i
    total = 0 # for문 안에서 변수를 초기화 한다면? 반복할때마다 초기화
    total += i

print('1 ~ 100까지의 합:', sum_value)
print('total:', total)
```

```python
# 1 ~ 100사이에서의 3의 배수의 합을 구하세요
```

```python
value = 0

for i in range(1, 101):
    if i % 3 == 0: 
        value += i
print(value)        
 
```

```python
# for - 1.원하는 횟수(범위)만큼 반복
#       2.저장된 값을 순서대로 하나씩 꺼내는 작업에도 사용
#list - 순서대로 저장한 값들의 묶음
for data in ['apple', 'cady', 'woter', '2026']:
    print(data)
```
