```python
# 문자열 데이터 : 문자열 저장, 저장된 문자열에 대한 함수(기능)
# str :  텍스트 데이터를 다루는 타입
# 시퀀스(순서) 타입이자 한번 생성되면 변하지 않는 객체(불변)
# 시퀀스 : 순서가 존재하는 데이터의 연속적인 모음
#       글자 순서가 달라지면 -> 다른 데이터. (순서 유지!!)
#       문자열도 글자 순서대로 번호(인덱스)가 붙음
#       인덱싱, 슬라이싱, 연결, 반복, in, len() 작업가능
```

```python
# str + str 연결
a = 'python'
b = 'good'

print(a + ' is ' + b)
```

```python
# str * int 반복
 print('*' * 30)
```

```python
# 인덱싱 : 한글자 꺼내기. 문자열[인덱스]
# 인덱스 : 0부터 정수로 순서대로 지정
str1 = 'hello world'
print('str1[0]:', str1[0]) # 첫번째 글자 가져오기
print('str1[4]:', str1[4])
# 음수 인덱스 : 역순으로 -1 부터 시작

print('str1[-1]:', str1[-1]) # 제일 마지막 글자 가져오기
print('str1[-11]:', str1[-11])
```

```python
# 슬라이싱(범위로 잘라내기) 문자열[strat :  end] start부터 end 앞에까지
text = 'python programming'

print(text[0:6]) # 0 ~ 5
print(text[:]) # 값 생략시 -> 전부 다
print(text[:6]) # start값 생략 -> 0부터(처음부터)
print(text[7:]) # end값 생략 -> 끝까지

# 문자열[start : end : step] start부터 end 앞에까지 step씩 건너뛰면서 가져옴
print(text[::3]) # 전체 문자열에서 3씩 증가한 인덱스 번호에 맞는 문자만 
print(text[::-1]) # 문자열 역순으로 만들기

print(text[-50:50]) # 범위에 맞지 않는 인덱스 -> 가능한 전체 범위 문자열 가져옴
print(text[-15:10])
print(text[7:-5])

# print(text[-50]) # 인덱싱은 없는 인덱스 사용 X : srting index out of range
```

```python
# 문자열의 길이 : 가지고 있는 문자 개수(공백도 포함)
# len(문자열)
# len() -1 -> 마지막 글자의 인덱스 번호

txt1 = 'python'
print(len(txt1))
print(txt1[len(txt1)-1]) # 제일 마지막 글자 가져오기
```

```python
# 문자열 내 특정 문자열 바꾸기
txt2 = 'ppthon'
# 1번 인덱스 글자를 'y' 로 바꾸자
# txt2[1] = 'y' #문자열은 바꿀 수 없다
txt2 = txt2[0] + 'y' + txt2[2:]
print(txt2)
```

```python
for s in txt2:
    print(s) # 문자열 처음부터 끝까지 한글자씩 꺼낼때
```
