```python
# 함수 : 기능
# 내장함수 : 파이썬이 기본적으로 제공하는 함수
#          len(), max() ... 함수()
# 문자열 내장함수 : 문자열.함수()
#                 저장된 문자열에 대한 작업
```

```python
# 문자열.count('찾을 문자열') : 찾을 문자열이 몇개 있는지 세어주는 함수
a = 'Pythonpython'
result = a.count('th')
print(f'th 는 {result}개 있습니다.')

result = a.count('p') # 대소문자 구별
print(f'p 는 {result}개 있습니다.')

result = a.count('zzz') # 없으면 0
print(f'zzz 는 {result}개 있습니다.')
```

```python
# find('찾는 문자열') : 0번 인덱스부터 찾아서 해당 문자열의 제일 처음 나온 위치 반환
# find('찾는 문자열', 시작 인덱스) : 시작인덱스부터 찾아서 해당 문자열의 위치 반환
# rfind('찾는 문자열') : 오른쪽부터 찾는 함수

a = 'Pythonpython'
res1 = a.find('th') # 2
print(res1)
res2 = a.find('y', 6)
print(res2)
res3 = a.rfind('y')
print(res3)

res4 = a.find('zzzz') # 없으면 -1 
print(res4)
```

```python
# index('찾을 문자열) : find()와 동일. 찾는 문자열이 없으면 에러(valueError)
res5 = a.index('y')
print(res5)
```

```python
# 대소문자 변환
# upper() : 대문자 변환
# lower() : 소문자 변환
# title() : 각 단어의 앞글자만 대문자로 변환
# capitalize() : 첫 문자를 대문자로 변환

# 긴 문자열 선언
# 쉼표없이 공백(줄바꿈, tab)으로만 구분해서 (문자열1 문자열2 ...) 선언하면
# 파이썬은 하나의 긴 문자열로 자동 연결함.
sentence = (
    "Python is a versatile programming language that is widely used in web development, "
    "data analysis, artificial intelligence, and automation, and many developers appreciate "
    "its clear syntax and readability because it allows them to focus more on solving problems "
    "rather than struggling with complex code structures."
)
print(sentence.upper(), end ='\n\n')
print(sentence.lower(), end ='\n\n')
print(sentence.title(), end ='\n\n')
print(sentence.capitalize(), end ='\n\n')

print('전체 글자 수',len(sentence))

print('p 개수: ', sentence.count('p'))
print('대소문자 상관 없이 p 개수: ', sentence.lower().count('p'))

print('Python 이라는 문자열로 시작: ', sentence.startswith('Python'))
print('Python 이라는 문자열로 끝: ', sentence.endswith('Python'))

# 공백 삭제
# strip()  좌우 공백 제거(문자열 내부의 공백은 제거 X) - 외부에서 읽어온 데이터에서 공백을 제거하는 용도로 사용
# rstrip() 오른쪽 공백 제거
# lstrip() 왼쪽 공백 제거

a = '     h e l l o       '
print(len(a))
res = a.strip()
print(f'공백 제거: {res}, 제거 후 길이{len(res)}, 원본 문자열은 그대로 {len(a)} ')
res = a.rstrip()
print(f'오른쪽 공백 제거: {res}, 제거 후 길이{len(res)}, 원본 문자열은 그대로 {len(a)} ')
res = a.lstrip()
print(f'왼쪽 공백 제거: {res}, 제거 후 길이{len(res)}, 원본 문자열은 그대로 {len(a)} ')

# replace('기존 문자열','새 문자열') 문자열 변경(공백 제거시에도 사용)
b = 'data1 data2 data3'

res = b.replace('data', 'value')
print(res)

res = b.replace('data', 'value', 1)
print(res)



res = b.replace(' ','') # 공백을 빈문자열로 변경(공백 제거)
print(res)

# islower() 문자열이 소문자로 이루어져있는지 여부 반환
# isupper() 문자열이 대문자로 이루어져있는지 여부 반환

b = 'HELLO'
print(b.islower())
print(b.isupper()) 

# isdigit() 숫자로 이루어진 문자열인지 여부 반환
# int로 변환하기 전에 검사하는 용도

c = '12345'
d = '1235.0'
print(f'c:{c.isdigit()}, d:{d.isdigit()}')

# input_data = input('정수 입력해')
# print(input_data.isdigit())
```

```python
# split('구분자') : 문자열을 지정한 구분자로 분리해 여러개의 문자열 생성 -> 리스트로 모아서 저장
# 구분자를 지정하지 않으면 기본 구분자로 분리(공백, /n, /t)
str1 = 'hello python world'
print( str1.split())
str2 = 'hell\pytho\torld'
print( str2.split())
```

```python
str3 = 'a,b,c,d,e'
print(str3.split())
print(str3.split(',')) # 구분자는 사람짐
```

```python
ss = input('날짜를 입력하세요(yyyy/mm/dd)')
# 입력한 날짜의 10년후 날짜를 출력하세요
input_date = ss.split('/') # 2026/10/06 -> ['2026, '10', '06']
year = int(input_date[0]) + 10

print('10년후 날짜', f'{year}년 {input_date[1]}월 {input_date[2]}일')
```

```python
#splitlines() : 행단위(\n)를 기준으로 분리해주는 함수
str4 = '''하나
둘
셋 넷\n다섯'''
print(str4)
print(str4.split())
print(str4.splitlines())    
```

```python
# 리스트 안에 있는 여러 개의 문자열 합치기
# join() : 리스트를 합쳐 하나의 문자열로 만들어주는 함수
#         1.묶어줄 구분자를 문자열로 지정
#         2.'구분자'.join(리스트)

s_list = ['a','b','c','d']
print(''.join(s_list))
print(','.join(s_list))
print('\n'.join(s_list))
```

```python
ss = '파이썬'
print(ss.center(10)) #숫자만큼 자릿수잡고 문자열을 가운데 배치
print(ss.center(10,'='))#숫자만큼 자릿수잡고 문자열을 가운데 배치, 빈자리 문자열로 채움
print(ss.ljust(10)) # 왼쪽에 붙여서 출력
print(ss.rjust(10)) # 오른쪽에 붙여서 출력
print(ss.zfill(10)) # 오른쪽에 붙여서 출력, 왼쪽 빈자리는 0으로 채움
```
