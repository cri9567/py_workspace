```python
# type hint

# 매개변수:타입
# 매개변수에 어떤 타입 값을 넣으라는 힌트

# def 함수명() -> 리턴타입 : 
# 함수가 끝나면 어떤 타입의 값을 반환하는지 알려주는 힌트
```

```python
def add(a:int = 0, b:int = 10) -> int:
    return a + b

print(add())
print(add(5, 1))


```

```python
def get_value(flag:bool) -> int | str:
    if flag:
        return 100
    else:
        return "백"

print(get_value(True))
print(get_value(False))            
```

```python
def get_list() -> list[int]:
    return [1,2,3,4]
  
```
