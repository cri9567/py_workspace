```python
# 패킹(packing) : 한 변수에 여러개의 테이터를 할당
# 언패킹(unpacking) : 묶여있는 여러개의 데이터를 각각의 변수로 반환
```

```python
t = 1, 2, 4 # tuple로 패킹
a, b, c = t # unpacking

print(t, a,b,c)
```

```python
list1 = ['a','b','c','d']
a,b,c,d = list1
print(a,b,c,d) # 리스트도 언패킹 가능
```

```python
a, b, c = list # 값의 수와 변수의 수가 일치 X -> error
```

```python
# * 값,값,값 -> 패킹
# * 자료구조 -> 언패킹
a,b, *c = list1
print(a,b,c)

numbers = [1, 15, 16]
print(*numbers) # 언패킹
```

```python
a = [1 ,2]
b = [3, 4]

result = [*a, *b] # 리스트 두개를 합칠때 사용
print(result)
```
