```python
#2d_list : 이차원 리스트
# 리스트 안에 리스트가 들어있느 형태
# 표 형태의 데이터를 다루거나 행렬(매트릭스) 연산시 사용

# 선언
# [[],[],[]]

# 검색
# 이차원리스트[인덱스][인덱스]
```

```python
# 5명의 학생 - kor, math, eng
kor_score = [49, 79, 20, 100, 80]
math_score = [43, 59, 85, 30, 90]
eng_score = [49, 79, 48, 60, 100]

a_class = [kor_score, math_score, eng_score]
print(a_class)
```

```python
# a_class 성적에서국 국어 성적 중 2번 학생의 성적을 출력
print(a_class[0]) # 1차원 리스트를 가져옴
print(a_class[0][2])
```

```python
# 2차원 리스트는 1차원 리스트의 주소값을 모아서 저장
math_score[0] = 1000 # 1차원 리스트 변경했지만
print(a_class) # 같은 주소이기 때문에 변경됨
```

```python
# a_class에서 math성적 중 0번학생의 성적을 100으로 
a_class[1][0] = 100
print(a_class)
```

```python
for i in [1,2,3,5]:
    print(i)
```

```python
# in 2차원 리스트 -> 1차원 리스트를 꺼내옴
total = []

for score in a_class: #[[kor],[math],[eng]]
    print(score)
    total += score

print('total : ', total)    
```

```python
total = 0
for score in a_class: #[[kor],[math],[eng]]
    for s in score:
        print(s, end=' ')
        total += s
    print()
print('total:', total)        
```

```python
# 이차원 리스트 개수 -> 가지고 있는 1차원 리스트의 개수
print(len(a_class))
```
