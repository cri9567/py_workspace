```python
# stack : 나중에 들어온 값이 먼저 나가는 구조 LIFO(순서 유지)
# 저장 -> 제일 뒤
# 꺼내기 -> 제일 마지막 데이터만 (삭제하면서)

stack = [1,2,3,4,5]
# 저장 : 제일 뒤
stack.append(10)
print('스택에 추가:', stack)

# 꺼내기
print('값 추출:', stack.pop())
print('값 추출 후:', stack)

print('값 추출:', stack.pop())
print('값 추출 후:', stack)
```

```python
# queue : 먼저 들어온 값이 먼저 나가는 구조(FIFO)
# 저장 -> 제일 뒤 
# 꺼내기 -> 제일 앞에 있는 데이터( 0번 인덱스)

q = [1,2,3,4,5]

q.append(10)
q.append(20)

print('10,20 추가 후:', q)

print(q.pop(0))
print(q)

print(q.pop(0))
print(q)

print(q.pop(0))
print(q)
```

```python
# 최근 검색어 취소
search_history = []
search_history.append('파이썬')
search_history.append('스택')
search_history.append('큐')

# 가장 최근 검색어 제거
s = search_history.pop()
# 제거된 검색어 출력
print(s)
# 남은 검색어 출력
print(search_history)
```

```python
queue = []

while True:
    print("\n1. 주문 접수")
    print("2. 주문 처리")
    print("3. 대기 주문 확인")
    print("0. 종료")

    menu = input("메뉴 선택: ")

    if menu == "1":
        order = input("주문 이름: ")
        queue.append(order)
        print("주문 접수 완료")

    elif menu == "2":
        # 주문 내역(리스트)에 주문이 있을때만 삭제 -> 없는데 삭제하면 에러
        if queue:
            order = queue.pop(0)
            print("주문 처리:", order)
        else:
            print("처리할 주문이 없습니다.")

    elif menu == "3":
        print("대기 주문:", queue)

    elif menu == "0":
        print("프로그램 종료")
        break

    else:
        print("메뉴를 다시 선택하세요.")
```
